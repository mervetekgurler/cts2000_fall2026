Now that we warmed up by counting punctuation, let's tackle the elephant in the room: **tokens**

You probably heard of tokens in the context of large language models, and we will get to that in the next few days, but for the purposes of this notebook, let's stick to an older definition of a token:

> Tokenization is the process of splitting text into _something_ - often words. **Each word is called a "token"** and a word such as "the" might adhere to multiple tokens of "the" within a text based on its capitalization, punctuation, etc.

The `.split()` method allows us to split the text based on some sort of separator. The default is blank and will split on the white spaces between words. As such, we are using the `.split()` method as our **tokenizer**.

## Step 4: Stopwords

[Stop words](https://en.wikipedia.org/wiki/Stop_words) are the most common words in a language, and may or may not add information about the content of the analysis.

There are a lot of ways to handle stopwords in a corpus. If your (English language) texts have over a million or so words, you can simply get rid of top 100-200 most frequent words, depending on the specifics of your corpus.

Otherwise, another very common method is to use lists of stopwords.

Similar to what we did with punctuation, we will use an external library to acquire a list of stopword, specificially designed for English

````python
import nltk
nltk.download("stopwords")
from nltk.corpus import stopwords
````

[NLTK](https://www.nltk.org/) stands for Natural Language Toolkit and is a Python library that provides tools for working with human language data (over 50 languages!), such as tokenizing text, removing stopwords, and analyzing word frequencies.

We will talk more about NLTK! For now, let's just take a quick look at their stopwords

````python
stopwords_en = stopwords.words("english")
print(stopwords_en)
````

````python
words_no_stop = []

for word in words:
    if word not in stopwords_en:
        words_no_stop.append(word)

print(words_no_stop)
````

````python
words_no_stop_counts = Counter(words_no_stop)
words_no_stop_counts.most_common(10)
````

## Step 5: Sentences

Another common way to approach text analysis is to start with sentences. If tokenization assumes that the smallest unit of analysis is a word, splitting text into sentences, assumes that the smallest unit of analysis is a sentence. 

Splitting into sentences is super useful for many syntax analysis. Sentences retain a lot more context than individual words. Let's look at this sentence from Letter 35

`The burying fields about it are certainly much larger than the whole city.`

`The burying fields` here refers to cemeteries, whereas in another context `burying` could be a verb. When we split the whole text into words, we inevitably flatten some of the meaning in these texts.

Luckily, there is another NLTK module for sentence tokenization

````python
nltk.download('punkt')
nltk.download('punkt_tab')
from nltk.tokenize import sent_tokenize
````

````python
sentences = sent_tokenize(letter_35)
````

````python
# Here is another way to do a fancy print statement

print(f"Number of sentences in letter 35: {len(sentences)}")
````

````python
# And another wat

counter = 1
for sentence in sentences:
    print(counter) 
    print(sentence)
    counter += 1
````

Printing it this way, we can see that the sentences have some line break issues. This is normal and doesn't really mean much, but let's clean it either way.

````python
letter_35 = letter_35.replace('\n', ' ')
````

````python
sentences = sent_tokenize(letter_35)
````

````python
counter = 1
for sentence in sentences:
    print(counter) 
    print(sentence)
    counter += 1
````

````python
# When we split it sentence by sentence, we can see that the sentences have some line break issues
# Sometimes analyzing the text can reveal formatting issues that need to be addressed, 
# especially when dealing with historical texts or texts extracted from formats like ePub.

# Let's fix the line breaks in the actual letter

letter_35_test = letter_35.replace('\n', ' ')

letter_35_sentences_test = sent_tokenize(letter_35_test)

print(f"Number of sentences in letter 35: {len(letter_35_sentences_test)}")

counter = 1
for sentence in letter_35_sentences_test:
    print(counter) 
    print(sentence)
    counter += 1
````
