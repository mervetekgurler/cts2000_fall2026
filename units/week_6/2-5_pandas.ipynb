## 1. What is a DataFrame?

A **DataFrame** is a two-dimensional table — like a spreadsheet — where each column holds a named field, and each row is one record. Pandas is designed around this object.

## 3. Selecting Columns, Rows, and Cells

### Single column

A single column on its own is called a **Series** — a one-dimensional labelled array.

You select a single column using *single brackets* and the name of the column that you want to select.

```python
languages['family']
```

```python
# Confirm the types
print(type(languages))             # DataFrame
print(type(languages['family']))   # Series
```

### Multiple columns

Two or more columns together return a **DataFrame**, think of it like a smaller version of your existing df.

You select multiple columns by using *double brackets* and the names of the columns you want to select.

```python
languages[["language", "family"]]
```

> 💡 **Why double brackets?**  
> `languages['family']` — single string → returns a **Series**  
> `languages[["language", "family"]]` — list of strings → returns a **DataFrame**

### Accessing one row

```python
# Get the row 6 (0-indexed)

languages[5]
```

> This does not work because square brackets on a DataFrame expect a column name, not a row number. Pandas looks for a column called 5, doesn't find it, and errors.

But of course, there are a lot of ways to access a single row!

### Slicing rows by position

```python
languages[0:5]
```

```python
# Rows 5 through 9 (like slicing a list)

languages[5:10]
```

```python
# Applying the slicing logic to row 6

languages[5:6]
```

```python
languages[5:6].shape
```

We get a dataframe with 1 row and 5 columns

### `.iloc` — integer-based indexing

`.iloc[rows, columns]` uses **integer positions** for both rows and columns.

```python
# Get the row 6 (0-indexed)

languages.iloc[5]
```

We can use `.iloc` to return non-consecutive rows. Pass in **integers** as a double list. 

```python
# Get the 1st, 5th, and 10th rows (0-indexed)

languages.iloc[[0, 4, 9]]
```

```python
# Rows 0–4, columns 0–2 (language, family, branch)
# Remember that the ending point is exclusive

languages.iloc[0:5, 0:3]
```

```python
# Stride: every 3rd language between rows 0 and 14

languages.iloc[0:14:3]
```

```python
# Negative indexing: the last 3 rows

languages.iloc[-3:]
```

```python
# Last row

languages.iloc[-1]
```

We can also pass a second interior list in `.iloc` to specify columns

```python
# First 3 rows, only language and family columns
languages.iloc[0:3, [0, 1]]
```

```python
# 1st, 5th, and 9th row, only language and L2 speakers
languages.iloc[[0, 4, 8], [0, 4]]
```

### `.loc` — label-based indexing

`.loc[rows, columns]` uses **row index labels** and **column names**. Usually clearer than `.iloc` when column names are readable.

```python
# 1st, 5th, and 9th row, only language and L2 speakers
languages.loc[[0, 4, 8], ['language', 'l2_speakers_m']]
```

```python
# Top 5, language and family

languages.loc[0:4,['language', 'family']]
```

> ⚠️ **`.loc` vs `.iloc`**  
> `.loc[0:4]` is **inclusive** on both ends → rows 0, 1, 2, 3, 4  
> `.iloc[0:4]` is **exclusive** on the right → rows 0, 1, 2, 3

On a side note, this would also work if your rows had custom labels, say if you were using unique ids for your datapoints in each row.

### Selecting single cells

You can use both `.loc` and `.iloc` to select single cells

```python
# Find the 11th language with most speakers
# hint: the dataframe is sorted based on total L1+L2 speakers

languages.iloc[10, 0]
```

```python
languages.columns
```

```python
# how many L1 speakers does this language have?

languages.loc[10, 'l1_speakers_m']
```

## 4. Boolean (conditional) subsetting

The most powerful way to filter rows is with a **boolean mask** — a Series of True/False values that you use to index the DataFrame.

Let's see how this plays out: 

**Find all the Indo-European languages in this df**

```python
# Step 1: create the mask
languages['family'] == 'Indo-European'
```

```python
# Step 2: use the mask to filter
ie = languages[languages['family'] == 'Indo-European']
print(ie.shape)
ie.head()
```

**Find all the *NON* Indo-European languages in this df**

```python
# Inequality

not_ie = languages[languages['family'] != 'Indo-European']
print(not_ie.shape)
not_ie.head(3)
```

### Combining conditions

Use `&` (AND) and `|` (OR). **Always wrap each condition in parentheses.**

```python
# Indo-European languages that are also Romance languages

ie_romance = languages[
    (languages['family'] == 'Indo-European') &
    (languages['branch'] == 'Romance')
]

print(ie_romance.shape)
ie_romance
```

```python
# Romance or Semitic languages

semitic_romance = languages[
    (languages['branch'] == 'Semitic') |
    (languages['branch'] == 'Romance')
]

print(semitic_romance.shape)
semitic_romance
```

### `.str` accessor — filtering on string content

For string columns, pandas offers a `.str` accessor that unlocks methods like `.contains()`, `.startswith()`, `.lower()`, etc. Very useful for text analysis.

```python
# Languages that contain 'Arabic' in their name
languages[languages['language'].str.contains('Arabic', case=False)]
```

## 5. Adding and modifying columns

### Creating a new column

We can create a new column by combining or transforming existing columns. This is very useful for storing transformations that we do with our data for later use.

```python
# Compute total speakers (L1 + L2) and add it as a new column
languages['total_speakers'] = languages['l1_speakers_m'] + languages['l2_speakers_m']
languages[['language', 'l1_speakers_m', 'l2_speakers_m', 'total_speakers']]
print(languages.shape)
languages.head()
```

### Renaming columns

We can rename our columns. Here, I am not sure if I am happy with `total_speakers`, it would match the rest of the col names better if it was `total_speakers_m` showing that it is in millions

```python
# Rename with a dictionary: {old_name: new_name}
# inplace=True modifies the DataFrame directly instead of returning a copy
languages.rename(columns={
    'total_speakers':   'total_speakers_m'
}, inplace=True)

languages.columns
```

### Deleting a column

Dropping a column with `.drop()` returns a copy by default, meaning that it does not modify the dataframe. We need to either create a new dataframe with fewer columns or use `inplace=True` argument to change the df.

```python
languages.shape
```

```python
languages.drop(columns=['family'])
```

```python
languages.shape
```

This will be handy for versioning on the fly and subsetting your dataframe

```python
languages_v2 = languages.drop(columns=['family', 'branch'])
languages_v2.head()
```

## 6. Counting, sorting, and grouping

### `.value_counts()` — frequency table for one column

```python
languages['language'].value_counts()
```

`value_counts()` by default groups unique items together

```python
languages["branch"].value_counts()
```

This means that we can use `value_counts()` to find the top 3 most frequent language family branches in our dataset

```python
languages["branch"].value_counts().head(3)
```

```python
# why not this?

languages["branch"].head(3).value_counts()
```

The order in which we chain functions matter! In the second example, we got the counts of different branches among the top 3 languages

### `.unique()` and `.nunuque()`

`unique()` returns the actual unique values, `nunique()` gives you the number of unique values in your data

```python
# How many unique language families are in our dataset?
languages['family'].nunique()
```

```python
# What are the unique language families?
languages['family'].unique()
```

### `.sort_values()` — sort by a column

```python
# Sort by total speakers, largest first
languages[['language', 'total_speakers_m']].sort_values('total_speakers_m', ascending=False)
```

<div style="background: rgba(167, 139, 250, 0.12); border-radius: 6px; padding: 0.6em 1em; margin: 1.5em 0; font-weight: 500;">
  Did anything change in this sorting?
</div>




No, because this data was already sorted by total number of speakers in Wikipedia/Ethnologue. We just left out the L1+L2 column to show you how this transformation works.

Let's sort it by the **number of L1 speakers**:

```python
# Sort by L1 speakers, smallest first
# to sort in ascending order, you can use the .sort_values() method without the ascending argument. It defaults to True
languages[['language', 'l1_speakers_m']].sort_values('l1_speakers_m', ascending=True)
```

Arabic (Modern Standard) is a special case — it has *no* L1 speakers because no one grows up speaking the standardised written form natively; speakers first acquire a regional dialect.

Do we have to use `.sort_values()` functions with numbers only?

```python
# Sort alphabetically by language name
languages.sort_values('language').head(10)
```

```python
# Sort alphabetically by family, then by total speakers within each family
languages.sort_values(['family', 'total_speakers_m'], ascending=[True, False])[['language', 'family', 'total_speakers_m']]
```

### `.groupby()` — split, apply, combine

`groupby()` splits the DataFrame into groups based on a column, **applies an aggregate function** to each group, then combines the results. Think of it as a pivot table.

```python
# Number of languages per family in our dataset
languages.groupby('family')['language'].count().sort_values(ascending=False)
```

Let's look at this in more detail, to understand what the code is doing

1. `.groupby('family')` -> groups our dataset by family
2. `['language'].count()` -> counts how many languages fall into each grouping
3. `.sort_values(ascending=False)` -> sorts the counts by descending order

```python
# Average L1 speakers per language family
languages.groupby('family')['l1_speakers_m'].mean().round(2).sort_values(ascending=False)
```

```python
# Total speakers (L1 + L2) per branch
languages.groupby('branch')['total_speakers_m'].sum().sort_values(ascending=False)
```

## 7. Missing Data

Real datasets almost always have gaps. Pandas represents missing values as `NaN` (Not a Number).

```python
# Check how many missing values per column
languages.isna().sum()
```

We have one missing value in the branch column, meaning that one of the 15 languages does not have a branch

```python
# Show only rows where any value is missing
languages[languages.isna().any(axis=1)]
```

This makes sense! Japanese is considered a language isolate

```python
# Drop rows where 'branch' is missing
languages_with_branch = languages.dropna(subset=['branch'])
print(languages_with_branch.shape)
languages_with_branch[['language', 'branch']]
```

```python
languages_with_branch.isna().sum()
```

## 9. Key points to remember

| Concept | Code |
|---|---|
| Load CSV | `pd.read_csv('file.csv')` |
| First N rows | `df.head(N)` |
| Shape | `df.shape` |
| Column names | `df.columns` |
| Data types | `df.dtypes` |
| Summary stats | `df.describe()` |
| Select column | `df['col']` (Series) |
| Select columns | `df[['col1', 'col2']]` (DataFrame) |
| Position indexing | `df.iloc[rows, cols]` |
| Label indexing | `df.loc[rows, cols]` |
| Boolean filter | `df[df['col'] == value]` |
| AND / OR | `(cond1) & (cond2)` / `(cond1) \| (cond2)` |
| String search | `df['col'].str.contains('text')` |
| New column | `df['new'] = ...` |
| Rename columns | `df.rename(columns={'old': 'new'})` |
| Count values | `df['col'].value_counts()` |
| Group + aggregate | `df.groupby('col')['other'].mean()` |
| Sort | `df.sort_values('col', ascending=False)` |
| Missing values | `df.isna().sum()` / `df.dropna()` / `df.fillna()` |
