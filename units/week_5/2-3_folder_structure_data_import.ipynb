```python
# "print" working directory
!pwd
```

We actually want the "data/" folder inside of the main "montagu" directory, so we can change this working directory path with the ["os" module](https://docs.python.org/3/library/os.html) to interact with your computer's operating system. 

```python
# Import the os module
import os

# Another way to check the cwd ("current" working directory) using the os module
os.getcwd()
```

Argh! Our default working directory is wherever we launched the notebook - in our case the "02_unit" folder inside the "units" folder. We want to access the "data" folder, which is two levels "up", inside of the main "montagu" directory. 

We can edit the file path to the correct directory. 

- We could type `os.chdir("../")` to go up one level into the "units" folder.
- We could type `os.chdir("../../")` to go up two levels into the "montagu" directory.  
- Or, we could move into the "data" directory in one line by typing `os.chdir("../../data")`

```python
os.chdir("../")
```

```python
# We include one ../ for each level up in the file structure
os.chdir("../../data")
```

# Import .txt file

Now that Python is looking in the correct location, we can pass in a single argument to the `open()` function - the name of the file! The `.read()` method from open states that the end of the file has been identified. 

```python
montagu = open("montagu_2006.txt")
```

```python
montagu = open("montagu_2006.txt").read()

# If your characters don't look quite right, try adding the argument encoding = "utf-8")
# frank = open("frankenstein.txt", encoding = "utf-8").read()
print(montagu)
```
