# Dictionary

A python dictionary is an [ordered (in new versions of python)](https://medium.com/junior-dev/python-dictionaries-are-ordered-now-but-how-and-why-5d5a40ee327f) collection of **key/value pairs - the key is the name of the data, the value is the data itself**. By calling the key, you can always find its corresponding value. Keys must be unique and immutable; strings and integers are commonly used as keys. The values can be anything, including lists, and even other dictionaries.

Each key/value pair is denoted by a colon `:` with the **key on the left side** and the **value on the right**. Each key/value pair within the dictionary is separate by a comma.

> NOTE: we have seen round parentheses `()` for functions, square brackets `[]` for indexing/slicing, but now note the use of curly braces `{}` to define a dictionary. 

````python
my_list = []
````

## Tuples

Tuples are similar to lists and arrays, but they are unchangeable ("immutable"). 

For example, we cannot change the output of 10 (the dividend) divided by 3 (the divisor) - by definition the quotient is 3 and the remainder is 1 this cannot be changed. 

> Although they appear similar to lists, note that tuples use round parentheses `()` to store information instead of square brackets.

````python
div = divmod(10,3)
print(div)
print(type(div))
````

````python
# Tuples use parentheses instead of square brackets

author = ("Toni", "Morrison", 1931)
author
````

````python
# indexing works the same way as lists

print(author[0])   # first name
print(author[-1])  # birth year
````

````python
# slicing too

author[:2]   # just the name parts
````

The key difference is that tuples are immutable

````python
# This works for a list:
my_list = ["Toni", "Morrison", 1931]
my_list[0] = "James" 

my_list
````

````python
# This breaks for a tuple:
author[0] = "James"    # TypeError!
````

````python
# convert tuple to a list

list(author)
````

````python
type(author)
````

````python
type(list(author))
````

Immutability is a very useful feature! We need a lot of things to be immutable so that we don't mistakenly edit them, like coordinates or publication years of works.

### Sets

A set is an unordered, unindexed collection of unique values — the key word being unique. It automatically drops duplicates.

Like dictionaries, sets have curly braces, but no key/value pairs

````python
# Remember our library from the very first example?

list_library = ["Catch-22", "The Brothers Karamazov", "Little Women", "Crime and Punishment", 
           "The Alchemist", "Rashomon", "Catch-22", "Pride and Prejudice"]
list_library
````

````python
set_library = set(list_library)
set_library
````

Main use of a set is to de-duplicate lists.

But you cannot index a set!

````python
# easy with a list
list_library[0]
````

````python
# not so much with a set
set_library[0]
````

Remember the definition! Sets are unordered and unindexed!

But they are super useful for checking membership (whether or not something is in a set)

````python
"Little Women" in set_library
````

````python
"Oliver Twist" in set_library
````

Oliver Twist was a later addition to our library, we can still bring him in.

Sets support math-style operations:

````python
new_books = {"Oliver Twist"}
type(new_books)
````

````python
# this is what it would look like as a dictionary
# title of the book: how many copies I have

new_books_dict = {"Oliver Twist": 5}
````

````python
type(new_books_dict)
````

````python
new_books_dict["Oliver Twist"]
````

````python
new_books["Oliver Twist"]
````

````python
updated_library = set_library | new_books
updated_library
````

````python
"Oliver Twist" in updated_library
````

These are the other math-like functions with sets

```python
A = {"fiction", "poetry", "biography"}
B = {"poetry", "drama", "biography"}

A & B   # intersection — what's in both: {"poetry", "biography"}
A | B   # union — everything: {"fiction", "poetry", "biography", "drama"}
A - B   # difference — in A but not B: {"fiction"}
```
