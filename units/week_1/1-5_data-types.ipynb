# Data Types

_Everything_ in Python has a type. Everything belongs to a class with certain properties that define it. This is important because this type determines what we can do to that variable/function/argument/thing. The most common types of data are float, integer, string, and logical. 

* **int**: Integers (e.g., `a = 2`).
* **float**: Decimal numbers (e.g., `a = 2.01`).
* **str**: Strings, which denotes text (e.g., `a = "2"` or `a = '2'`).
* **bool**: Boolean values, which are either `True` or `False` (e.g., `a = True`).

## Type Conversion

Types can get confusing. For instance, we can write a number as either an integer or a string. Python treats these differently, even if to us the value is the same:

```python
a = '3'
b = 3

b - a
```

Even though our intention is to do numeric subtraction, the type of `a` is a string, which results in an error.

If we could convert this to an integer, the operation will work. 

We can do this with **type conversion**. The `int()` function will convert the input to an integer:

```python
int(a)
```

```python
type(int(a))
```

```python
b - int(a)
```

There are other type conversion functions.

- `str()` converts a variable to a string.
- `float()` converts a variable to a float. 

⚠️ **Warning:** If the value cannot be converted to that type, the function will return a `ValueError`. Run the cell below.

```python
int('Netherlands')
```

In the above case, the error means that **non-numeric characters** cannot be interpreted as a number.

## Strings have methods! 

You've been using **functions** like `print()` and `type()`, to carry out common tasks with data and variables.

A **method** is a special type of function: one that belongs to a **particular type of object**, like a string or an integer. Methods allow you to do different things with different objects.

For instance, we can use a method to turn a string variable into lowercase or uppercase. These lowercase and uppercase methods don't exist for integers, though. That's why we call them methods instead of functions – and why we access them in a different way.

You can access (and recognize) methods through **dot notation**. It looks like this: `variable.method()`

Let's look at the built-in method `upper()`, which can be applied to a string-type variable:

```python
# This is called "dot notation"
status.upper()
```

```python
continent.lower()
```

## Indexing, Slicing, and Striding

### Indexing

Indexing is a way to access just a part of some variable by referring to the position of that part. 

> NOTE: Instead of round parentheses `()` like we see after the names of functions, indexing/slicing/striding all use square brackets `[]`

```python
pangram = "The quick brown fox jumps over the lazy dog"
print(pangram)
```

```python
type(pangram)
```

If we just want the first letter of this sentence, we could just type the number 1 in square brackets to refer to the letter in the first position, right?  

**Wrong! :)**

```python
pangram[1]
```

Python is a ["zero-indexed"](https://www.quora.com/What-is-the-meaning-of-%E2%80%9CPython-is-a-zero-indexing-language-unlike-R-where-indexing-starts-at-one%E2%80%9D) language, meaning counting starts at zero. Therefore, we would need to type 

```python
pangram[0]
```

### Slicing

What if we want to index a range of values? We can slice a variable to get multiple indices by placing a colon inside of our square brackets `[:]`. The index before the colon is _included_ in the slice, but the number after the colon is _excluded_.

```python
# Slice only the first four characters:
# print(pangram[0:4])
pangram[0:4]
```

```python
# Slice only the last five characters:
# print(pangram[-5:])
pangram[-5:]
```

```python
# Slice the 9th through the 12th characters
# print(pangram[8:12])
pangram[8:12]
```

### Striding

Include a second colon in your slice to specify an interval/step:

```python
# Slice 0:12, but by every third letter
print(pangram)
pangram[0:12:3]
```
