## 1. Thinking about visualization

### What is a visualization for?

> The power of a graph is its ability to enable one to take in the quantitative information, organize it, and see patterns and structure not readily revealed by other means of studying the data.

\- Cleveland and McGill, 1984

> A good graphic realizes two basic goals: It **presents** information, and it allows users to **explore** that information.

\- Alberto Cairo

> A data visualization should only be beautiful when beauty can promote understanding in some way without undermining it in another. Is beauty sometimes useful? Certainly. Is beauty always useful? Certainly not.

\- Stephen Few

These quotes share one idea: a visualization exists to help someone **understand** something. A figure that looks impressive but doesn't help a reader see a pattern, or worse, makes them see one that isn't there, has failed, however beautiful it is.

### How we read a chart

In their 1984 paper *Graphical Perception*, William Cleveland and Robert McGill tested how accurately people can read different kinds of visual information. Their main finding: we judge some things much more accurately than others. We are very good at comparing **positions** along a common scale and **lengths**, and much worse at comparing **areas**, **angles** and **colour intensity**.

Look at these circles. Which is bigger, L or M?

<img src="../../img/circles.png" alt="Circles of different sizes, labelled with letters" width="500" style="background-color: white; padding: 12px; border-radius: 6px;">

For circles of very different sizes, the comparison is easy: A is smaller than B. But for L and M, which are almost the same size, it's hard to tell. Now look at the same values as bars:

<img src="../../img/bars.png" alt="The same values shown as bars" width="500" style="background-color: white; padding: 12px; border-radius: 6px;">

Now it's clear which is larger. This is why bar charts are so common: **length** is one of the things we read most accurately.

What about two series at once? Imagine we have English and Math test scores for each student. If we **stack** them, we can compare the total scores:

<img src="../../img/two-series-0.png" alt="Stacked bars of English and Math scores" width="500" style="background-color: white; padding: 12px; border-radius: 6px;">

We can see that Y and L had the highest totals. But which student scored highest in Math? That's hard to tell, because the Math parts of the bars don't start from the same line: they sit on top of the English scores. Cleveland and McGill call this a **non-aligned scale**. One solution is to put the two scores on opposite sides of the axis:

<img src="../../img/two-series-1.png" alt="English and Math scores on opposite sides of the axis" width="500" style="background-color: white; padding: 12px; border-radius: 6px;">

Now it's easy to see that R did very well in Math. But it's harder to compare the totals. There is no single best chart: **it depends on what you want your reader to see.**

💡 This is also why we were careful with **colour** in the heatmaps: colour intensity is one of the things we read least accurately, which is why a few very high values can make everything else look the same.

### Start with a question, not a chart

The most common mistake in data visualization is choosing a chart first: *"I'll make a network!"* or *"I'll make a map!"*. Instead, start by writing down, in one sentence, **what you want your reader to see**. The question then tells you which chart to use.

| If your question is about... | Try... | We did this with... |
|---|---|---|
| how values are **distributed** | histogram, boxplot | word counts in the Gothic novels |
| **comparing** categories | bar chart | most common words |
| **change over time** | line chart, small multiples | - |
| **where** things are | map | places in Montagu's letters |
| how two categories **combine** | heatmap | sending locations vs. places mentioned |
| **connections** between people or things | network | characters in *She Stoops to Conquer* |
| the **structure** of a text | annotated text | Stanza's output |

A map is only the right choice if *location* matters to your question. A network is only the right choice if *connections* do.

### Every choice is an argument

A visualization is never just "the data". It's the data plus dozens of decisions, and each one shapes what a reader sees. We've made many of these decisions this term:

- **What to include:** we mapped only the places Montagu mentioned often, and drew the act-by-act networks with only the main cast.
- **How to define things:** we decided that two characters *interact* when one speaks right after the other.
- **How to group things:** we decided who belongs to the Hardcastle household and who is a visitor from London.
- **How to scale things:** we capped the heatmap's colour scale, so a few large values wouldn't hide everything else, and we used a log scale for really big numbers.

None of these choices is wrong. But your reader should be able to **see** that you made them. Say what you included and what you left out, label everything, and explain any unusual scale.

A few things to watch out for, because they can mislead a reader even by accident:

- **Bar charts should start at zero.** A bar's length is what we compare, so cutting off the bottom of the axis makes small differences look huge.
- **Averages hide the spread.** Two groups can have the same mean and look nothing alike. When you can, show the individual data points.
- **Counts depend on size.** A longer text, or a bigger archive, will have more of almost everything. Remember the raw vs. relative frequency discussion in the statistics notebook. We'll see this again very soon.

### Make it readable for everyone

- **Think of appropriate titles and captions for your figures**
- **Label your axes**, including units: *"Share of articles (%)"*, not *"share"*.
- **Make text big enough** to read on a slide, from the back of a room.
- **Check your figure on a white background.** Many of us use dark mode, but your figure will end up on a white page or slide. We'll come back to this when we save figures.

## 2. Our data: the Reuters news corpus

Our dataset for this notebook comes with `nltk`: the **Reuters corpus**. It contains **10,788 short news articles** sent out on the Reuters newswire in 1987. Each article has been labelled with one or more of **90 categories**, mostly about business and the economy: company earnings, acquisitions, currencies, and commodities like grain, oil, coffee and sugar.

It's a classic dataset in text analysis research, especially for **text classification**: teaching a computer to guess an article's category from its words. For us, it has three advantages:

- It's **clean, born-digital text**, so we don't have to fight with OCR errors.
- It's **big enough** to use for word vectors: over 1.3 million words.
- It downloads with one line of code, like the stopword list.

Like every dataset, it also has limits, and we'll run into some of them. Keep asking yourself: **what can this data tell us, and what can't it?**

If you would like to read more about this dataset, here is the original README file from 1997 written by [David D. Lewis from AT&T Labs.](https://kdd.ics.uci.edu/databases/reuters21578/README.txt)

We are accessing this corpus through NLTK. You can read more about [NLTK's corpora](https://www.nltk.org/book/ch02.html) and specific [methods](https://www.nltk.org/howto/corpus.html) they developed for working with these texts.

````python
import nltk
nltk.download("reuters")
nltk.download("stopwords")
from nltk.corpus import reuters, stopwords
````

### Loading the corpus

The corpus works like the other `nltk` corpora we've used. Each article has a **file id**, and we can ask for its categories and its text:

````python
first_id = reuters.fileids()[0]

print(first_id)
print(reuters.categories(first_id))
print(reuters.raw(first_id)[:400])
````

Notice that the file id starts with `test/`. The corpus is split into a **training** set and a **test** set, because it was made for training classifiers: you teach a model with the training articles, then check how well it does on the test articles it has never seen.

Let's put everything into a DataFrame, with one row per article:

````python
rows = []
for fileid in reuters.fileids():
    text = html.unescape(reuters.raw(fileid))     # turn codes like &lt; back into <
    rows.append({
        "fileid": fileid,
        "split": fileid.split("/")[0],              # "training" or "test"
        "categories": reuters.categories(fileid),  # a list: an article can have several
        "title": text.split("\n")[0].strip(),       # the first line is the headline
        "text": text,
        "words": len(text.split()),
    })

news = pd.DataFrame(rows)
news.head()
````

Let's unpack this:

`html.unescape(...)` -> the original files write some characters as **HTML codes**: `<` is written `&lt;`, and `&` is written `&amp;`. Company names often appear as `&lt;Exxon Corp>`, which is hard to read. `html.unescape` turns the codes back into the real characters.

`"categories": reuters.categories(fileid)` -> this column holds a **list** in each row, because one article can belong to several categories. We'll need to handle that carefully.

`text.split("\n")[0]` -> the text before the first line break is the headline.

````python
print(len(news), "articles")
print(news["words"].sum(), "words")
````

````python
# let's see what all the categories are
print(len(reuters.categories()), "categories")
reuters.categories()
````

There are 90 categories and they seem to be very specific.

````python
# we can see which article has which categories
reuters.categories('training/9865')
````

````python
# we can also see which articles are in a given category
print("there are", len(reuters.fileids('barley')), "articles in barley category")
reuters.fileids('barley')
````

````python
# We can also see how the categories are distributed over our articles
news["categories"].str.len().value_counts().sort_index().rename_axis("number of categories").rename("articles")
````

There are 9160 articles with only one category, so 84.9% of all articles. But 30 articles are tagged with 7 or more categories!

## 3. From a question to a figure

For each question, we'll start with the quickest possible chart, then improve it step by step, just like we did with the networks.

### Question 1: What is the news about?

This is a question about **comparing categories**, so a bar chart is a good start. But first, we need to count the categories, and each article has a *list* of them. `.explode()` gives each item in a list its own row:

````python
article_categories = news.explode("categories")
article_categories[["fileid", "categories"]].head()
````

An article with three categories now appears in three rows, once for each category. Now we can count them:

````python
print("There are now", len(article_categories), "articles, instead of 10,788")
````

2,540 additional rows! These are NOT new articles. The increase is caused by the way that we changed the structure of our dataframe. Instead of one row representing one article, each row represents *one article + one category*

We did this to visualize the distribution of different categories across the corpus:

````python
category_counts = article_categories["categories"].value_counts()
category_counts.plot.bar(); # the ; hides the "<Axes: ...>" text above the chart, useful for quick pandas based visualizations
````

<div style="background: rgba(167, 139, 250, 0.12); border-radius: 6px; padding: 0.6em 1em; margin: 1.5em 0; font-weight: 500;">
  What can you learn from this chart? What can't you read?
</div>

This is a typical **first** chart: technically correct, but almost unreadable.

- There are **90 bars**, and the labels are too small and sideways to read.
- Two categories are so big that all the others are squashed at the bottom. This is a **long tail**, like Zipf's law in the statistics notebook.
- The labels are **codes**: what are *earn*, *acq* or *dlr*? A reader who doesn't know the dataset can't tell.

Maybe we can do some small adjustments?

````python
category_counts.plot.bar(figsize=(18, 5), fontsize=9);
````

`figsize=(18, 5)` -> the side of the figure in inches figsize=(width, height)

`fontsize=9` -> font for the labels

Sure, maybe we see a bit more of the data now, but this does not invalidate the other issues, namely the long tail and the abbreviations.

Let's keep only the 15 biggest categories, turn the bars sideways so the labels are easy to read, and translate the codes into plain words.

````python
top_15 = category_counts.head(15)
print(top_15)
````

````python
# plain-language names for the 15 biggest categories
# modified from the dataset documentation: https://archive.ics.uci.edu/dataset/137/reuters+21578+text+categorization+collection

category_names = {
    "earn": "Earnings", "acq": "Mergers & acquisitions",
    "money-fx": "Money/foreign exchange", "grain": "Grain", "crude": "Crude oil",
    "trade": "Trade", "interest": "Interest rates", "ship": "Shipping", "wheat": "Wheat",
    "corn": "Corn", "dlr": "U.S. dollar", "money-supply": "Money supply",
    "oilseed": "Oilseed", "sugar": "Sugar", "coffee": "Coffee",
}
````

````python
# what share of all articles are about earnings or acquisitions?
earn_or_acq = news["categories"].apply(lambda cats: "earn" in cats or "acq" in cats).mean()

fig, ax = plt.subplots(figsize=(9, 6))
ax.barh([category_names[c] for c in top_15.index], top_15.values,
        color=["#0072B2" if c in ("earn", "acq") else "#A7B0BC" for c in top_15.index])
ax.invert_yaxis()                      # the biggest bar at the top

ax.set_title(f"{earn_or_acq:.0%} of the articles are about company earnings or acquisitions",
             loc="left", fontsize=14, fontweight="bold")
ax.set_xlabel("Number of articles")
ax.spines[["top", "right"]].set_visible(False)

plt.show()
````

Let's unpack the new parts:

`fig, ax = plt.subplots(figsize=(9, 6))` -> creates a **figure** (the whole image) and an **axes** (the chart inside it). Instead of `plt.title()` and `plt.xlabel()`, we now write `ax.set_title()` and `ax.set_xlabel()`. It does the same thing, but it's clearer which chart we're changing, and it's how most examples online are written.

`ax.barh(...)` -> a **horizontal** bar chart. Long labels are much easier to read sideways than at an angle.

`[category_names[c] for c in top_15.index]` -> the plain-language name for each category code.

`color=["#0072B2" if c in ("earn", "acq") else "#A7B0BC" ...]` -> blue for the two categories the title talks about, grey for the rest. **Colour points the reader** to the part of the chart that matters.

`ax.invert_yaxis()` -> by default, the first bar is drawn at the bottom. Flipping the axis puts the biggest one at the top, where we read first.

`lambda cats: "earn" in cats or "acq" in cats` -> for each article's list of categories, `True` if it includes *earn* or *acq*. `.mean()` of a column of `True`s and `False`s is the **share** that are `True`.

The title now states the finding, and it comes from the data: if the data changed, the number would change with it.

**Is there another way to visualize these categories? Perhaps a *pie chart*?**

Pie charts are everywhere, and you may be tempted to use one in your final project. Before you do, let's see what happens when we draw our categories as a pie.

````python
# the 15 biggest categories, plus everything else together as one slice
#top_15 = category_counts.head(15)
pie_data = pd.concat([top_15, pd.Series({"all other categories": category_counts.iloc[15:].sum()})])

fig, ax = plt.subplots(figsize=(8, 8))
ax.pie(pie_data, labels=[category_names.get(c, c) for c in pie_data.index],
       autopct="%1.0f%%", startangle=90, counterclock=False)
ax.set_title("Categories in the Reuters corpus?", loc="left", fontsize=14, fontweight="bold")
plt.show()
````

This pie says earnings reports are 30% and Mergers & acquisitions are 18%. But in Question 1, we found that they cover about 59% of all articles. Which is right?

````python
print(category_counts.sum(), "category labels")
print(len(news), "articles")
````

There are 13,328 category labels, but only 10,788 articles, because an article can have several categories. So the slices don't add up to the articles. They add up to the labels, and an article about grain and wheat is counted in two slices. A pie chart says "these slices are parts of one whole", and here there is no such whole. That's why the percentages don't match.

A pie chart only works when every item belongs to exactly one slice, so the slices really add up to 100% of something. For example, every article has either one category or several:

````python
n_categories = news["categories"].str.len()
one_or_more = pd.Series({
    "one category": (n_categories == 1).sum(),
    "two or more categories": (n_categories > 1).sum(),
})

fig, ax = plt.subplots(figsize=(6, 6))
ax.pie(one_or_more, labels=one_or_more.index, autopct="%1.0f%%", startangle=90, counterclock=False,
       colors=["#0072B2", "#D55E00"], wedgeprops=dict(edgecolor="white", linewidth=1),
       textprops=dict(fontsize=12))
ax.set_title(f"{one_or_more.iloc[0] / one_or_more.sum():.1%} of articles have only one category",
             loc="left", fontsize=14, fontweight="bold")
plt.show()
````

`ax.pie(values, labels=..., autopct="%1.0f%%")` draws the pie. autopct writes each slice's percentage on it, with no decimals.

`startangle=90, counterclock=False` starts the first slice at 12 o'clock and goes clockwise.

`wedgeprops=dict(edgecolor="white", linewidth=1)` adds thin white lines between the slices, so they're easier to tell apart.

`pd.concat([...])` sticks the top 10 and the "all other categories" slice together into one Series.

### Question 2: How long are the articles?

This is a question about a **distribution**, so let's start with a histogram:

````python
news["words"].hist();
````

Almost all the articles are in the first two bars, so we can't see much. The distribution is **skewed**: most articles are short, but a few are very long and stretch out the axis. Narrower bins will show more detail:

````python
fig, ax = plt.subplots(figsize=(10, 4))
sns.histplot(data=news, x="words", binwidth=5, color="#0072B2", ax=ax)
ax.set_xlim(0, 600)
ax.set_xlabel("Length of article (words)")
ax.set_ylabel("Number of articles")
plt.show()
````

`sns.histplot(data=news, x="words", ...)` -> this is how `seaborn` works: we give it the **table** (`data=`) and the **name of the column** (`x=`). We don't have to pull the column out ourselves.

`ax=ax` -> tells `seaborn` to draw on **our** chart, so we can keep adjusting it with `matplotlib`.

`binwidth=5` -> each bar covers 5 words: articles with 0 to 4 words, 5 to 9 words, and so on. Instead of choosing the *number* of bins, we choose how *wide* each one is.

`ax.set_xlim(0, 600)` -> we zoom in on the first 600 words, where almost all articles are. That's a choice the reader should know about.

Now look at the very left of the chart: there's a small group of articles with almost **no words** at all, separated from the rest by a gap. Before we show this chart to anyone, let's look at them:

````python
news.sort_values("words")[["fileid", "categories", "words", "text"]].head()
````

````python
very_short = news[news["words"] < 15]
print(len(very_short), "articles with fewer than 15 words")
````

These "articles" contain nothing but a **date**, like *26-FEB-1987*. Their text was lost or never included, but they're still in the corpus, with categories. They're not real articles, so we'll **remove** them, and say so.

💡 Always look at the extremes of your data, the shortest, the longest, the biggest and the smallest, before you trust a chart.

````python
news = news[news["words"] >= 15].copy()
print(len(news), "articles left")
````

**Let's change the number of bins in our histogram**

Bins are like intervals in which to break up the distribution of data. We can change the number of bins that we are splitting our data into and the best way to find the correct number is to try a few.

Note that this is different from changing the **bin width** to 5 like we did earlier. The number of bins is how many bars the histogram has. You say "25 bins", and Python divides the range into 25 equal pieces. The bin size, or bin width, is how wide each bar is. You say "5 words per bin", and Python works out how many bars that takes.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for bins, ax in zip([5, 25, 100], axes):
    ax.hist(news["words"], bins=bins, range=(0, 800), color="#0072B2", edgecolor="white")
    ax.set_title(f"{bins} bins", loc="left", fontsize=13, fontweight="bold")
    ax.set_xlabel("Length of article (words)")
    ax.spines[["top", "right"]].set_visible(False)

axes[0].set_ylabel("Number of articles")
fig.suptitle("The same data, three different bin numbers", x=0.01, ha="left", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
````

- 5 bins: far too few. Almost everything is in the first bar, and the shape of the distribution disappears.
- 25 bins: the overall shape appears: a peak around 50 words, then a long tail.
- 100 bins: much more detail, including a tiny bar on the far left, just above zero.

We can also compare a few different bin width options:

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for width, ax in zip([100, 25, 5], axes):
    ax.hist(news["words"], bins=range(0, 801, width), color="#0072B2", edgecolor="white", linewidth=0.5)
    ax.set_title(f"Bins {width} words wide", loc="left", fontsize=13, fontweight="bold")
    ax.set_xlabel("Length of article (words)")
    ax.spines[["top", "right"]].set_visible(False)

axes[0].set_ylabel("Number of articles")
fig.suptitle("The same data, three bin widths", x=0.01, ha="left", fontsize=15, fontweight="bold")
plt.tight_layout()
plt.show()
````

- 100 words wide: just a falling staircase. The peak around 50 words is hidden in the first bar.
- 25 words wide: the overall shape appears: a peak, then the long tail.
- 5 words wide: the most detail, including the very tall bars around 20 to 30 words, the short earnings reports, and the sudden drop after about 110 words.

### Smoothing a histogram: KDE

A histogram puts the data into separate bins. A **kernel density estimate (KDE)** draws a smooth curve instead, to show the general shape of a distribution. In seaborn, we add one with `kde=True`:

````python
fig, ax = plt.subplots(figsize=(10, 4))
sns.histplot(data=news, x="words", binwidth=10,
             color="#9CC3E4", edgecolor="white", linewidth=0.5,     # the data, as light bars
             kde=True, line_kws={"linewidth": 1.5}, ax=ax)          # the shape, as a curve
ax.lines[0].set_color("#D55E00")                                    # the curve in orange, on top

ax.set_xlim(0, 800)
ax.set_title("Length of articles, with a smoothed curve", loc="left", fontsize=14, fontweight="bold", pad=15)
ax.set_xlabel("Length of article (words)")
ax.set_ylabel("Number of articles")
ax.spines[["top", "right"]].set_visible(False)
plt.show()
````

The curve shows the **overall shape** nicely: a peak around 50 words, and a long tail to the right. It's a bit like a histogram with no bins. But look at what it smooths away:

- **The tallest bars**, around 20 to 30 words, almost disappear. Those are many very short earnings reports, a real feature of this corpus, and the curve flattens them into its peak.
- **The sudden drop** after about 110 words becomes a gentle slope.

So smoothing is a **choice** too, like the number of bins. A curve is easier to look at, and it's useful for comparing the shapes of several distributions on one chart. But it hides exactly the kind of detail we just found by looking closely at the bars. Use it when the general shape is what matters, and always check it against the histogram.

We can always add some more information as long as it conveys something useful and analytical:

### Measuring a relationship: correlation

The scatterplot suggests there's no strong relationship between a category's size and the length of its articles. Can we put a number on that? The most common measure is the **correlation coefficient**, written *r*. It's always between −1 and 1:

- **r = 1**: a perfect positive relationship. When one number goes up, the other always goes up too, along a straight line.
- **r = −1**: a perfect negative relationship. When one goes up, the other always goes down.
- **r = 0**: no straight-line relationship at all.

As a rough guide, values closer to 0 than about ±0.3 are usually called **weak**, around ±0.5 **moderate**, and above ±0.7 **strong**.

The most common version is **Pearson's correlation**. Here's the formula:

$$r = \frac{\sum (x - \bar{x})(y - \bar{y})}{\sqrt{\sum (x - \bar{x})^2 \cdot \sum (y - \bar{y})^2}}$$

It looks scary, but it's just a few steps. Let's calculate it ourselves, with $x$ as the number of articles and $y$ as the median length:

````python
x = by_category["articles"]
y = by_category["median_words"]

# step 1: how far is each category from the average, on each axis?
x_distance = x - x.mean()
y_distance = y - y.mean()

# step 2: multiply the two distances for each category, and add them up
top = (x_distance * y_distance).sum()

# step 3: divide by the spread of each variable, so r always ends up between -1 and 1
bottom = ((x_distance ** 2).sum() * (y_distance ** 2).sum()) ** 0.5

r = top / bottom
print(f"r = {r:.2f}")
````

**pandas** can do all of this in one line. Let's check that we get the same answer:

````python
x.corr(y)
````

Both give **r = −0.19**: a **weak negative** relationship. Bigger categories tend to have slightly shorter articles, but only slightly, which matches what we saw in the scatterplot.

Pearson's correlation measures **straight-line** relationships, and it's easily pulled around by extreme values. Our data has both problems: a long tail of category sizes, and a few huge categories like earnings. There are two common ways to deal with this.

**Spearman's correlation** works with the **ranks** instead of the values: the biggest category becomes 1, the second biggest 2, and so on. A huge category can then only be "first", not "50 times bigger than the rest".

````python
x.corr(y, method="spearman")
````

We could also use the **logarithm** of the number of articles, the same scale as our scatterplot's x-axis:

````python
import numpy as np
np.log10(x).corr(y)
````

| Measure | r |
|---|---|
| Pearson | −0.19 |
| Spearman (ranks) | −0.18 |
| Pearson, log scale | −0.33 |

All three are negative, so they agree on the **direction**. But depending on our choice, the relationship is either weak or moderately weak. That's not a mistake: each measure answers a slightly different question. What matters is that you choose one for a reason, and **say which one you used**.

For example, if I wanted to report any of these correlations, I would edit the caption like so: *"Figure 3. Bigger categories tend to have slightly shorter articles (Pearson's r = −0.19), but the smallest categories vary the most."*

### Question 3: Which categories appear together?

About 15% of the articles have more than one category. Which categories tend to go together? This is a question about how two categories **combine**, so let's use a **heatmap**, like the one for Montagu's sending locations and the places she wrote about.

For every pair of categories, we count how many articles have both:

````python
top_10 = list(top_10)
together = pd.DataFrame(0, index=top_10, columns=top_10)

for cats in news["categories"]:
    cats_in_top_10 = [c for c in cats if c in top_10]
    for a, b in itertools.product(cats_in_top_10, repeat=2):
        together.loc[a, b] += 1

together.columns = together.index = [category_names[c] for c in top_10]
together
````

`pd.DataFrame(0, index=top_10, columns=top_10)` -> an empty table full of zeros, with one row and one column for each category.

`itertools.product(cats_in_top_10, repeat=2)` -> every **pair** of categories in an article's list, including each category paired with itself. An article about grain and wheat gives us (grain, grain), (grain, wheat), (wheat, grain) and (wheat, wheat).

So the **diagonal** of the table, where a category meets itself, is just the number of articles in that category. Let's draw it:

````python
fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(together, annot=True, fmt="d", cmap="Blues", ax=ax)
plt.show()
````

<div style="background: rgba(167, 139, 250, 0.12); border-radius: 6px; padding: 0.6em 1em; margin: 1.5em 0; font-weight: 500;">
  Which cells catch your eye first? Are they the interesting ones?
</div>

The darkest cells are on the **diagonal**, and the darkest of all is earnings reports meeting itself. That's just the size of each category, which we already know from Question 1. Because those numbers are so big, they take up the whole colour scale, and the pairs we actually care about all look pale. It's the same problem as in the Montagu heatmap, where Vienna's 19 mentions washed out everything else.

The fix: **hide the diagonal**, so the colour scale only covers the pairs.

````python
# hide the diagonal: a category with itself is just the category's size
hide_diagonal = pd.DataFrame(False, index=together.index, columns=together.columns)
for name in together.index:
    hide_diagonal.loc[name, name] = True

fig, ax = plt.subplots(figsize=(9, 7))
sns.heatmap(together, mask=hide_diagonal, annot=True, fmt="d", cmap="Blues",
            linewidths=0.5, cbar_kws={"label": "Articles with both categories"}, ax=ax)

ax.set_title("Grain, wheat and corn almost always come together", loc="left", fontsize=14, fontweight="bold")
plt.setp(ax.get_xticklabels(), rotation=45, ha="right")     # tilt the labels, and line them up with their columns
plt.show()
````

Let's unpack this:

`mask=hide_diagonal` -> a table of `True` and `False` with the same shape as our data. Wherever it says `True`, seaborn leaves the cell empty.

`annot=True, fmt="d"` -> writes the number in each cell. `fmt="d"` means "as a whole number".

`linewidths=0.5` -> thin white lines between the cells, which make the grid easier to read.

`cbar_kws={"label": ...}` -> a label for the **colour bar** on the right, so the reader knows what the colours mean.

`plt.setp(ax.get_xticklabels(), rotation=45, ha="right")` -> tilts the column labels, and anchors each one at its right end (`ha` is short for *horizontal alignment*), so it ends right under its column.

Now the real patterns stand out:

- **Grain, wheat and corn** go together. But look closer: wheat and corn almost never appear *without* grain. The categories are **nested**: wheat and corn are kinds of grain, and whoever labelled the articles added both. That's a choice built into the dataset, not something about the news.
- **Foreign exchange and interest rates** often appear together, which makes sense: both are about money and central banks.
- **Earnings reports** hardly combine with anything.

### Question 4: What words does each category use?

So far, we've counted articles. Now let's look **inside** the text. Which words are most common in each category? We want to compare six categories, and putting them all in one chart would be crowded. Instead, we'll draw **small multiples**: the same small chart, repeated for each category, side by side. We did this with the act-by-act networks.

First, we count the words in each category, leaving out stopwords, numbers and very short words:

````python
stop_words = set(stopwords.words("english"))
stop_words.add("said")      # "said" appears in almost every news article, so it tells us little

def count_words(texts):
    counts = Counter()
    for text in texts:
        words = re.findall(r"[a-z]+", text.lower())
        counts.update(w for w in words if w not in stop_words and len(w) > 2)
    return counts

six_categories = ["earn", "acq", "crude", "grain", "money-fx", "coffee"]

word_counts = {}
for category in six_categories:
    texts = news[news["categories"].apply(lambda cats: category in cats)]["text"]
    word_counts[category] = count_words(texts)

word_counts["coffee"].most_common(10)
````

`stop_words.add("said")` -> we add one word to the stopword list. In news, *said* is everywhere (*"the company said"*), so it would top every chart. Removing it is a choice, and different choices give different charts.

`re.findall(r"[a-z]+", text.lower())` -> splits the text into words made only of letters, in lower case, which also drops numbers.

````python
fig, axes = plt.subplots(2, 3, figsize=(15, 8))

for category, ax in zip(six_categories, axes.flatten()):
    top_words = pd.DataFrame(word_counts[category].most_common(8), columns=["word", "count"])
    sns.barplot(data=top_words, x="count", y="word", color="#0072B2", ax=ax)
    ax.set_title(category_names[category], loc="left", fontsize=13, fontweight="bold")
    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.spines[["top", "right"]].set_visible(False)

fig.suptitle("The most common words in six categories", x=0.01, ha="left", fontsize=16, fontweight="bold")
plt.tight_layout()
plt.show()
````

Let's unpack this:

`fig, axes = plt.subplots(2, 3, figsize=(15, 8))` -> a grid of 6 charts: 2 rows, 3 columns.

`axes.flatten()` -> turns the 2 x 3 grid into one list of 6 charts, so we can loop over them together with our 6 categories using `zip`.

`sns.barplot(data=top_words, x="count", y="word", ...)` -> with the words on the y-axis, seaborn draws horizontal bars, and keeps the order of our table, so the most common word is at the top.

`fig.suptitle(...)` -> a title for the whole figure, above all six charts.

`plt.tight_layout()` -> adjusts the spacing so the charts and labels don't overlap.

Notice that each chart has its **own** x-axis: the numbers for earnings reports are much bigger than for coffee, because there are many more earnings reports. That's fine here, since we're comparing *which* words appear, not *how many*. But if you wanted to compare counts across charts, you'd need `sharex=True` or relative frequencies.

Word clouds break almost every rule we've seen:

- **They ask us to compare areas.** Remember Cleveland and McGill: we read size much less accurately than length. And a word's size depends on more than its count: long words like *production* take up much more space than short words like *pct*, even when they're less common.
- **The size isn't even proportional.** By default, the `wordcloud` library sizes words partly by their **rank**, not only by their count, so a word that appears twice as often isn't drawn twice as big.
- **Colour, position and direction mean nothing.** Some words are vertical, some horizontal, and the colours are random. But readers will look for meaning in them anyway: is the purple word important? Are words close together related?
- **There are no numbers.** We can't tell whether *oil* appears 200 times or 2,000.

### Putting tables in your notebook

Sometimes, you don't need a chart at all. Charts are good at showing **patterns**: what's bigger, what's rising, what stands out. Tables are better when the reader needs the **exact numbers**, or wants to look up one particular value. A good rule: if you only have a handful of numbers, or if the precise values matter more than the overall shape, use a table.

Our boxplot in Question 2 showed that earnings reports are the shortest articles and trade stories the longest. But how long, exactly? How many articles is each box based on? A table can answer that:

````python
length_table = (top_10_articles.groupby("category")["words"]
                .agg(["count", "median", "max"])
                .sort_values("median")
                .rename(columns={"count": "Articles",
                                 "median": "Median length (words)",
                                 "max": "Longest article (words)"}))
length_table.index.name = "Category"

length_table.style.format("{:,.0f}")
````

````python
# # optionally, save it as a csv
# length_table.to_csv("length_table.csv")
````

Let's unpack this:

`.agg(["count", "median", "max"])` -> calculates several statistics at once for each group: how many articles, the median length, and the longest article.

`.sort_values("median")` -> sorts the rows in the same order as the boxplot, from shortest to longest, so the reader can compare the two. As with charts, alphabetical order would hide the ranking.

`.rename(columns={...})` -> gives the columns names a reader understands, **with units**, just like axis labels.

`length_table.index.name = "Category"` -> names the first column.

`.style.format("{:,.0f}")` -> shows every number as a whole number, with thousands separators: *3,935* instead of *3935*, and *52* instead of *52.0*. A table full of decimals is hard to read, so round to what the reader needs. `.style` only changes how the table **looks**: the numbers inside `length_table` stay the same.

💡 For your final project, you can copy a table like this into your paper. `length_table.to_csv("length_table.csv")` saves it as a file you can open in Excel or Google Sheets, and paste from there into your document.

The same rules apply to tables as to charts:

- **Give it a title or caption** that says what it shows.
- **Name the columns clearly**, with units.
- **Round** to the precision your reader needs.
- **Sort** the rows in a meaningful order.
- **Keep it short.** If your table has 90 rows, it's probably a chart, or a table of the top 10.

## Key points to remember

| Idea | What it means |
|---|---|
| Start with a question | write what the reader should see, then choose the chart |
| Position and length | what we read most accurately; area, angle and colour much less so |
| Every choice is an argument | what you include, define, group and scale shapes what a reader sees |
| Raw vs. relative | when groups differ in size, compare shares, not counts |
| Know your archive | what survived, was digitized, sampled and labelled shapes every chart |
| Log scales | spread out a long tail, but say so in the axis label |
| Small multiples | the same small chart repeated, instead of one spaghetti chart |
| Other people's analysis | topics, places and people in a dataset are often model predictions |
| Titles state findings | *"Half of the stories ran in 5 newspapers or fewer"*, not *"Reprints per story"* |
| `fig, ax = plt.subplots()` | create a figure, then change it with `ax.set_...()` |
| `seaborn` | `data=` the table, `x=`, `y=` the column names, `ax=ax` to keep adjusting |
| `plt.savefig(...)` | before `plt.show()`, with `dpi=300, bbox_inches="tight", facecolor="white"` |
